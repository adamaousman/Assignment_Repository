```python
#without list comprehension
data=[1,2,3,4,5]
for i in data:
    print(i**2)

print(data)

#line of code
```

```python
#with list comprehension print square of each element
data=[1,2,3,4,5]
newData=[i**2 for i in data ]
print(newData)
```

```python
#print only event number from data in a new list
data=[23,12,9,7,26,80,44]
newData=[i for i in data if i%2==0]
print(newData)
```

```python
#print only odd number from data in a new list
data=[23,12,9,7,26,80,44]
newData=[i for i in data if i%2==1]
print(newData)
```

```python
# if mark>=60 pass else Fail
marks=[60,80,89,65,23,87,12,40]
result=["Pass" if i>=60 else "Fail" for i in marks ]
print(result)
```

```python
#need to store same data in new data list but in upper case
names=["priya","rahul","pawan","john","suman"]
result=[n.upper() for n in names]
print(result)
```

```python
#store same list without space
names=[" priya "," rahul "," pawan "," john","suman "]
result=[i.strip() for i in names]
print(result)
```
