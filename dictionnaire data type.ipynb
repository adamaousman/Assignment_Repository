```python
#duplicate key not allowed
#data={key:value}
data={"name":"mohit", "age":23, "college":"invertis", "name":"amit"} #the second name replaces the 1st value
print(data)
```

```python
#mutable
data={"name":"mohit", "age":23, "college":"invertis", "name":"amit"} #the second name replaces the 1st value
data["name"]="sumit"
print(data)
```

```python
#valid keys
data={1:"amit", "age":23 , 2.3:"version"}
print(data)
```

```python
#invalid keys
data={[1,2]:"version"} #can not use key as muatable data type 
print(data)
```

```python
#empty dict
data={}
print(data)
data=dict()
print(data)
```

```python
#accessing values
data={"name":"mohit", "age":23, "college":"invertis" } 

#method1 : using key
print(data["name"])

#method1 : using get () method
print(data.get("name"))
print(data.get("version")) #it's because the key is not in the dictionnary
```

```python
#adding new method into dictionnary
data={"name":"mohit", "age":23, "college":"invertis" } 
data["course"]="python programming"
print(data )
```

```python
#updating values
data={"name":"mohit", "age":23, "college":"invertis" } 
data.update({"name":"amit","age":27})
print(data )
```

```python
#methods of dictionnary

#len ()
data={"name":"mohit", "age":23, "college":"invertis" } 
print(len(data))
print(data )
```

```python
#keys()
data={"name":"mohit", "age":23, "college":"invertis" } 
print(data.keys())
print(data )
```

```python
#values()
data={"name":"mohit", "age":23, "college":"invertis" } 
print(data.values())
print(data )
```

```python
#items()
data={"name":"mohit", "age":23, "college":"invertis" } 
print(data.items())
print(data )
```

```python
#items()
data={"name":"mohit", "age":23, "college":"invertis" } 
print(data.pop("age"))
print(data)
```

```python
#copy()
data={"name":"mohit", "age":23, "college":"invertis" } 
newData=data.copy()
print(data)
print(newData)
```

```python
#CLEAR()
data={"name":"mohit", "age":23, "college":"invertis" } 
data.clear()
print(data)
```

```python
#membership operator
data={"name":"mohit", "age":23, "college":"invertis" } 
print("age" in data)
print("course" in data)
```

```python
#membership operator
data={"name":"mohit", "age":23, "college":"invertis" } 
print("age"  not in data)
print("course"  not in data)
```
