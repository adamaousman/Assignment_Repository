```python
#exo:3
products={"laptop":10, "mouse":3,"keyboard":0,"monitor":7,"printer":2}

for name, stock in products.items():
    if stock == 0:
        print(name, "out of stock")
    elif stock <=5 :
            print(name, "low stock")
    else:
             print(name, "available stock")
            
            
```

```python
#exo:4
employees={"rahul":{"salary":60000,"departement":"it"},
          "amit":{"salary":40000,"departement":"hr"},
          "priya":{"salary":70000,"departement":"it"},
          "neha":{"salary":45000,"departement":"finance"}}

for name , data in employees.items():
    salary=data["salary"]
    departement=data["departement"]

    if (salary >= 50000):
        if departement=="it":
            print(departement,"senior it employee")
        elif departement=="hr":
            print(departement,"senior hr employee")
        else:
            print(departement,"senior  employee")
    else: 
        print(departement,"junior  employee")
```
