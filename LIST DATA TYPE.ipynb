```python
#LIST DATA TYPE 
data=[10,20,30,40,50]
print("before change :",data[3])
data[3]=60 #list is mutable 
print("after  change :",data[3])
print(data[-1]) #negative indexing starts for last element
print(data[-2])
```

```python
studentrecord=["soniya",23,"java",2000.00, True] #heterogeneous type
print(studentrecord[0]) #indexing
print(studentrecord[0:3]) #slicing
```

```python
studentnames=["amit","soniya","pawan","amit","raj","john"] #duplicate elements
print(studentnames[0])
print(studentnames[3])
print(studentnames[:]) #print entire list
print(studentnames) #print entire list
```

```python
#Methods of list

studentnames=["amit","soniya","pawan","amit","raj","john"] 
#len()
print(len(studentnames))

#append
studentnames.append("priya")
print(studentnames)

studentnames.append(["sumit","yash"]) #adds the complete object as one element
print(studentnames)

#insert()
studentnames.insert(0,"rohan")
print(studentnames)
studentnames.insert(5,"adama")
print(studentnames)

#extends()
studentnames.extend(["gloria","paul","blessing"])
print(studentnames)

#remove()
studentnames.remove("amit")
print(studentnames)
studentnames.remove("sumit") #ValueError: list.remove(x): x not in list
print(studentnames)

#pop()

            
```

```python
data=["amit","soniya","pawan","amit","raj","john"] 
print(data.pop())
print(data)
print(data.pop(2))
print(data)
```

```python
#Methods of list

studentnames=["amit","soniya","pawan","amit","raj","john"] 
#clear
studentnames.clear("amit")
print(studentnames)
```

```python
#Methods of list
#index()
studentnames=["amit","soniya","pawan","amit","raj","john"] 
print(studentnames.index("amit"))
print(studentnames.index("amit",2))
```

```python
#Methods of list
#count()
studentnames=["amit","soniya","pawan","amit","raj","john"] 
print(studentnames.count("amit"))
```

```python
#sort ()

data=[54,65,34,45,20,78]
data.sort() #ascending order
print(data)
data.sort(reverse=True) #descending order
print(data)
```

```python
#reverse
data=[54,65,34,45,20,78]
data.reverse() 
print(data)
```

```python
#copy
data=[54,65,34,45,20,78]
newdata=data.copy() 
print(newdata)
```
