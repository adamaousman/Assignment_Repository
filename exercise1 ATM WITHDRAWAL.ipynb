```python
#ATM WITHDRAWAL PROGRAM
balance= 100000
correct_pin= 4671

pin=int(input("enter the pin"))

if pin==correct_pin :
    amount=int(input("enter the withdrawal amount"))

    if amount % 100 == 0:
        if amount <= balance :
           balance= balance -amount
           print("withdrawal succesful take your card and collect your cash")
           print("your new balance is : ", balance )
        else:
         print("insufficient balance")
    else:
      print("amount must be multiple of 100")
else:
    print("incorrect pin") 



        
```
