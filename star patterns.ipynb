```python
num=5
row=1
while(row<=num):
    #print number of spaces
    space=1
    while(space<=num-row):#5-1=4
        print(" ",end=" ")
        space+=1
        #print number of stars
    star=1
    while(star<=2*row-1):#(2*1)-1=1
        print("*",end=" ")
        star+=1
    print(end="\n")
    row+=1
```
