```python
#PROGRAM THAT TAKES MARKS OF 3 SUBJECTS

mark1=float(input("enter mark of maths"))
mark2=float(input("enter mark of physics"))
mark3=float(input("enter mark of biology"))

if mark1 >= 40 and mark2 >= 40 and mark3 >= 40 :
    total=mark1+mark2+mark3
    percentage=(total/300)*100
    if percentage >= 75 :
        print("Distinction")
    elif percentage >= 60 :
        print("first division")
    elif percentage >= 50 :
        print("second division")
    else :
        print("pass")
else:
    print("Fail")
```
