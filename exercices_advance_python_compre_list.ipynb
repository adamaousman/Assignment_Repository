```python
#Q1. Find numbers greater than 50 and store all the numbers into new list.
numbers = [20, 55, 30, 80, 45, 90, 10]
N_numbers=[i for i in numbers if i>50]
print(N_numbers)
```

```python
#Q2. Find numbers between 20 and 50 and store all the numbers into new list.
numbers = [10, 25, 35, 60, 45, 15, 50, 70] 
N_numbers=[i for i in numbers if i>=20 and i<=50]
print(N_numbers)
```

```python
#Q3. Find numbers divisible by both 3 and 5 and store all the numbers into new list.
numbers = [10, 15, 20, 30, 45, 50, 60, 75]
N_numbers=[i for i in numbers if i%3==0 and i%5==0]
print(N_numbers)
```

```python
#Q4. Find names having more than 4 characters and store result into new list.
names = ["Amit", "Rahul", "Neha", "Rohit", "Sneha"]
n=[i for i in names if len(i)>4 ]
print(n)
```

```python
#Q5. Find names starting with "A" and store result into new list.
names = ["Amit", "Rahul", "Ankit", "Neha", "Arjun", "Rohit"]
n=[i for i in names if i.startswith("A") ]
print(n)
```

```python
#06. Extract vowels from a string and store result into new list.
text = "programming"
vowels="a,e,i,o,u"
n=[char for char  in text if char in vowels  ]
print(n)
```

```python
#Q7. Extract digits from a string and store result into new list.
data = "Python123Programming456"
n=[i for i in data if i.isdigit()]
print(n)
```

```python
#Q9. Replace negative numbers with 0 and store result into new list
numbers = [10, -5, 20, -8, 30, -2]
result=[ex if ex>0 else 0 for ex in numbers]
print(result)
```

```python
#Q10. Give a 10% discount to products costing $1000 or more, otherwise give 5% and store result into new list
prices = [500, 1200, 800, 2000, 1500]
result=[ex*0.90  if ex>=1000  else ex*0.95 for ex in prices]
print(result)
```

```python
#Q12. Employee Management:
#Find employees earning more than 750,000.
employees = [
{"name": "Amit", "salary": 45000},
{"name": "Rahul", "salary": 60000},
{"name": "Neha", "salary": 75000},
{"name": "Rohit", "salary": 3500}
]
result=[info['name'] for info in employees if info['salary'] >50000]
print(result)
 
```

```python
#Q17. Find names of employees from the IT department.
employees = [
{"name": "Amit", "department": "IT"}, 
{"name": "Rahul", "department": "HR"},
{"name": "Neha", "department": "IT"}, 
{"name": "Rohit", "department": "Sales"}
]
result=[ ex['name'] for ex in employees if ex["department"]=="IT"]
print(result)
 
```

```python
#Q18. Find employees who earn more than 760,000 and belong to IT.
employees = [
{"name": "Amit", "department": "IT", "salary": 55000},
{"name": "Rahul", "department": "HR", "salary": 70000},
{"name": "Neha", "department": "IT", "salary": 80000},
 {"name": "Rohit", "department": "IT","salary": 45000}
 ]
result=[ex['name'] for ex in employees if ex['department']=="IT" and ex['salary']>=60000]
print(result)
```
