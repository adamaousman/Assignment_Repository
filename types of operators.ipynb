```python
#relational /comparision operators in python . relational operators are used to compare two values .
#the result of every relational operator is always a boolean value :true or false

x=10
y=20
print(x==y) #equal to == operator checks whether two values are equal.
print(x>y) #greater than > operator checks whether the left-hand value is greater than the right-hand value.
print(x<y) # less than < operator checks whether the left-hand value is less  than the right-hand value.
print(x>=y) #greater than or equal to >= operator cheks two conditions :
                                                                     #left-hand value is greater than the right-hand value.
                                                                     # or both value are equal
print(x<=y)  #less than or equal to >= operator cheks two conditions :
                                                                     #left-hand value is less than the right-hand value.
                                                                     # or both value are equal
print(x!=y)  # not equal to != operator checks whether two values are different .
```

```python
#Logical and operator in python:
#The and operator is used to combine two  or more conditions . and= BOTH conditions must be true .
x=10
y=20
print(x<y and y>x)
print(x<y and x==y)
print(x>y and x<y)
print(x>y and x==y)
```

```python
#Logical and operator in python:
#The and operator is used to combine two  or more conditions . or= at least one conditions must be true .
x=10
y=20
print(x<y or y>x)
print(x<y or x==y)
print(x>y or x<y)
print(x>y or  x==y)
```

```python
#Membership  operators in python 
#Membership  operators are used to check whether a particular value exists inside a collection or sequence.
#python provides two main membership operators : 1). in 2). not in 
data=[10,20,30,40,50]
print(10 in data) # the operator in checks whether a particular value exists inside a collection or not.
print(100 in data) # if the value is present  >> True and if the value is not present >> False
```

```python
#not in operator
#the not in operator checks whether a value does not exist inside a collection 
data=[10,20,30,40,50]
print(10 not in data) # if the value is present  >> False
print(100  not in data) # if the value is not present >> True
```

```python
x=10  #python creates an integer object containing 10
y=20  # now y refers to the integer object 20.

print(id(x))
print(id(y))

y=30 # The variable y is now referring to the object 30.
print(id(y)) #so the ID printed for y changes because y is now referring to a different object 30.

z=20 # in typical python implementations, especially CPython , small integers such as 20 are commonly reused /cached.
     #soz can refer to the same 20 object that was previously referenced by y.

print(id(z)) #giving the same ID that y had when y was referring to 20 .
```
