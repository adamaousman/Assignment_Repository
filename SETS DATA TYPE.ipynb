```python
#remove duplicate element
data={10,23,43,22,65,78,43,23,10}
print(data)
```

```python
#do not allow indexing
data={10,23,43,22,65,78}
print(data)
print(data[0])
print(data[0:6])
```

```python
#do not maintain order (it's unorder)
data={10,23,43,22,65,78}
print(data)
```

```python
#sets functions
add()
```

```python
#add() method
data={10,23,43,22,65,78}
data.add(45)
print(data)
```

```python
#update() method
data={10,23,43,22,65,78}
data.update([20,30,40])
print(data)
```

```python
#remove() method
data={10,23,43,22,65,78}
data.remove(78)
print(data)
data.remove(55) # it will raise error because 55 doesn't exist in the set
print(data)
```

```python
#discard()
data={10,23,43,22,65,78}
data.discard(55)
print(data)

```

```python
#pop()
data={10,23,43,22,65,78}
data.pop() #65
print(data)
```

```python
#clear()
data={10,23,43,22,65,78}
data.clear() 
print(data)
```

```python
#create empty set by using also 
data=set()
print(data)
```

```python
#copy()
data={10,23,43,22,65,78}
newData=data 
print(newData)

#USING COPY ()
copydata=data.copy()
print(copydata)
```

```python
#union()
A={10,20,30}
B={30,40,50}
print(A.union(B))
print(A|B) #WE CAN ALSO USE UNION OPERATOR

#intersection()
print("----------------------")
print(A.intersection(B))
print(A&B)

#difference()
print("----------------------")
print(A.difference(B))
print(A-B)
```
