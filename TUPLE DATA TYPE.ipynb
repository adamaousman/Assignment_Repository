```python
#tuple 
data=(10,20,30,40,50)
print(type(data))

```

```python
data2=10,20,30,40,50
    
print(type(data2))
```

```python
data3=(10) # we can not create tuple with single element 

print(type(data3))
```

```python
data4=(10,)  # we  create tuple with single element by using the value with coma separator

print(type(data4))
```

```python
data=(10,20,30,40,50)
print(data[2]) # indexing
print(data[-1]) #negative indexing
```

```python
data=(10,20,30,40,50,30)
print(data) # duplicate elements
```

```python
data=(10,20,30,40,50)
data[0]=88 # 'tuple' object does not support item assignment (SO TUPLE IS IMMUTABLE)
```

```python
#TUPLE PACKING
studentdata="aman","mcai ",78.65
print(type(studentdata))
#TUPLE UNPACKING
name,course,avgMarks=studentdata
print(name)
```

```python
record="amit","mca",23,True #tuple packing
name,course,age,att=record #tuple unpacking
name,*ab,att=record
print(name)
print(ab)
```

```python
#len()
data=(10,20,30,40,50)
print(len(data))
```

```python
#count()
data=(10,20,30,40,50,30)
print(data.count(30))
#sum()
print(sum(data))
print(min(data))
print(max(data))
print(data.index(10))
```
